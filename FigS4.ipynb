```r
seuobj = readRDS('/data/input/Files/ResultData/Notebook/liuyuyang/Mouse_Hippo/T321_T333_SCT_merge_bayes_anno.rds')

date()
plasma <- viridis::viridis(10, direction = 1, option = "C")

coord <- data.frame(stereo_1 = seuobj$coor_x_ad2, stereo_2 = seuobj$coor_y_ad2)
seuobj[["stereo"]] <- SeuratObject::CreateDimReducObject(embeddings = as.matrix(coord), key = "stereo_", assay = "Spatial")
```

```r
for (i in c('T323','T333') ){
    seu = subset(seuobj,sampleid == i)
    seu = subset(seu,scc.annoL1 %in% unique(seu$scc.annoL1)[grepl('HIP|SUBd',unique(seu$scc.annoL1) )] )
    p = ggplot(seu@meta.data,aes(as.numeric(coor_x_ad2), as.numeric(coor_y_ad2), fill= scc.annoL1))+ 
  geom_tile(width = 1, height = 1)+
  theme_classic()+
  coord_fixed()+ #theme(legend.position="None")+
  scale_fill_manual(values =cols)+
    geom_segment(aes(x = min(seu$coor_x_ad2) - 30, y = min(seu$coor_y_ad2) + 1, 
                     xend = min(seu$coor_x_ad2) - 30 + 20, yend = min(seu$coor_y_ad2) + 1),
                 color = 'black', size = 3)
    ggsave(paste0(i,'.area_anno.HIP.jpg') ,p,width = 5,height = 5)
    ggsave(paste0(i,'.area_anno.HIP.pdf') ,p,width = 5,height = 5)
    
    p = FeaturePlot(seu , features = "Etv1",reduction = "stereo",pt.size = 0.3)+scale_colour_gradientn(colours = plasma )+
     # labs(subtitle  = i)+
    theme(legend.position="None",
          axis.line = element_blank(), axis.text = element_blank(), axis.ticks = element_blank(), axis.title = element_blank())+coord_fixed()
    ggsave(paste0(i,'.Etv1.HIP.jpg') ,p,width = 4,height = 4)

    p = FeaturePlot(seu,features = "Fn1",reduction = "stereo",pt.size = 0.3) +
    scale_colour_gradientn(colours = plasma,limits = c(0, 4),oob = scales::oob_squish) +
    theme(
        legend.position = "none",
        axis.line = element_blank(),
        axis.text = element_blank(),
        axis.ticks = element_blank(),
        axis.title = element_blank()) +coord_fixed()
    ggsave(paste0(i,'.Fn1.HIP.jpg') ,p,width = 4,height = 4) 
    
    print(i)
}
```
